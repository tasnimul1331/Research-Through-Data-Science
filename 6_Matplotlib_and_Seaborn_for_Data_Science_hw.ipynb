```python
## Task 1: Basic Matplotlib (3 plots)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from google.colab import drive
drive.mount('/content/drive')
import warnings
warnings.filterwarnings('ignore')
```

```python
gym = pd.read_csv('/content/drive/MyDrive/data_set/gym_membership.csv')
```

```python
# 1. Bar chart of gym membership types
gym['abonoment_type'].value_counts().plot(kind='bar')
plt.title("Gym Membership Types")
plt.xlabel("Membership Type")
plt.ylabel("Number of Members")
plt.show()
```

```python
# 3. Scatter plot: Age vs visits per week
plt.scatter(gym['Age'], gym['visit_per_week'])
plt.title("Age vs Gym Visits Per Week")
plt.xlabel('Age')
plt.ylabel('visits')
plt.show()
```

```python
# 2. Histogram of time in gym

plt.hist(gym['avg_time_in_gym'],bins=40)
plt.title('Avarage Time In Gym')
plt.show()
```

```python
# 1. Box plot: Gender vs Age
sns.boxplot(data=gym, x='gender', y='Age')
plt.title('Gender vs Age')
plt.show()
```

```python
# 2. Count plot: Days per week
plt.figure(figsize=(30, 8))
sns.countplot(data=gym, x='days_per_week')
plt.xticks(rotation=90)
plt.title('Days per week')
plt.show()
```

```python
from matplotlib.typing import LineStyleType
# 3. Bar plot: Average visits by membership type
plt.grid(True, which='both',linestyle='--')
sns.barplot(data=gym, x='abonoment_type', y='visit_per_week',palette=['lightblue', 'salmon'])
plt.title('Average visits by membership type')
plt.show()
```

```python
fig,axes = plt.subplots(2, 2, figsize=(10, 8))
axes[0, 0].bar(gym['abonoment_type'].value_counts().index, gym['abonoment_type'].value_counts())
axes[0, 0].set_title('Gym Membership Types')
axes[0, 0].set_xlabel('Membership Type')
axes[0, 0].set_ylabel('Number of Members')

axes[0,1].hist(gym['avg_time_in_gym'],bins=40)
axes[0,1].set_title('Avarage Time In Gym')


male_age = gym[gym['gender'] == 'Male']['Age']
female_age = gym[gym['gender'] == 'Female']['Age']

axes[1,0].boxplot([male_age,female_age])
axes[1,0].set_xticklabels(['Male', 'Female'])
axes[1,0].set_title('Gender vs Age')

x_part =gym['abonoment_type']
y_part =gym['visit_per_week']
axes[1,1].bar(x_part, y_part)
axes[1,1].set_title('Average visits by membership type')



plt.tight_layout()
plt.show()
```
